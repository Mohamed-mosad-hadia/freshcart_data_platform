# Health Check status and Time 

```python
import requests

url = "http://localhost:8000/health"
headers = {"X-API-Key": "freshcart-fx-key"}

try:
    response = requests.get(
        url,
        headers=headers,
        timeout=10,
    )
    response.raise_for_status()
    data = response.json()
    print(data)
    print(f"Status Health: {data['status']}")
    print(f"Time Health: {data['time']}")

except requests.exceptions.RequestException as e:
    print(f"API request failed: {e}")
```

# The Currencies

```python
url = "http://localhost:8000/v1/currencies"
headers = {"X-API-Key": "freshcart-fx-key"}

try:
    response = requests.get(
        url,
        headers=headers,
        timeout=10,
    )
    response.raise_for_status()
    data = response.json()
    print(data)
    print(f"Base Currencies: {data['base']}")
    print(f"Supported Currencies: {data['currencies']}")
    print(f"Pegged Currencies: {data['pegged']}")
    print(f"SAR Pegged Currencies: {data['pegged']['SAR']}")
    print(f"AED Pegged Currencies: {data['pegged']['AED']}")
    print(f"EGP floating Currencies: {data['floating']}")
    

  
except requests.exceptions.RequestException as e:
    print(f"API request failed: {e}")
```

# The Rates API 

```python
import requests

url = "http://localhost:8000/v1/rates"
headers = {"X-API-Key": "freshcart-fx-key"}

try:
    response = requests.get(
        url,
        params={"date": "2023-01-01"},
        headers=headers,
        timeout=10,
    )
    response.raise_for_status()
    data = response.json()
    print(data)
    
    print("USD per EGP:", 1 / data["rates"]["EGP"])
    print("USD per SAR:", 1 / data["rates"]["SAR"])
    print("USD per AED:", 1 / data["rates"]["AED"])
except requests.exceptions.RequestException as e:
    print(f"API request failed: {e}")
```
